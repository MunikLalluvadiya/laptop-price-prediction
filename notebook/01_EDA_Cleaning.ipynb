EDA AND DATA CLEANING

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/laptops.csv", encoding="latin-1")
```

```python
df["GPU"] = df["GPU"].fillna("Integrated")
```

```python
df["Storage type"] = df["Storage type"].fillna("SSD")
```

```python
df.isnull().sum()
```

```python
df["Screen"].dtype
```

```python
df["Screen"] = df["Screen"].fillna(df["Screen"].mean())
```

```python
df.describe()
```

```python
df = df[df["Storage"] != 0].reset_index(drop=True) 
df.shape
```

```python
df.info()
```

```python
df["Touch"].value_counts()
```

Save Clean Data After EDA and Cleaning

```python
df.to_csv("../data/laptops_cleaned.csv", index=False, encoding="utf-8")
```
